```python
import csv
import sys

import matplotlib.pyplot as plt
```

```python
circuit = "IQP25"

match circuit:
    case  "SU5":
        istate = "0"
        ostate = "0"
        samples = ["25","25"]
        threads = ["16","16"]
    case  "SU15":
        istate = "0"
        ostate = "0"
        samples = ["32","32"]
        threads = ["128","128"]
    case  "SU25":
        istate = "0"
        ostate = "0"
        samples = ["36","36"]
        threads = ["128","128"]
    case  "IQP5":
        istate = "0"
        ostate = "0"
        samples = ["25","25"]
        threads = ["128","128"]
    case  "IQP15":
        istate = "0"
        ostate = "0"
        samples = ["33","33"]
        threads = ["128","128"]
    case  "IQP25":
        istate = "0"
        ostate = "0"
        samples = ["36","36"]
        threads = ["128","128"]
    case  "RND8":
        istate = "0"
        ostate = "0"
        samples = ["26","26"]
        threads = ["64","64"]
    case  "RND10":
        istate = "0"
        ostate = "0"
        samples = ["27","27"]
        threads = ["128","128"]
    case  "RND12":
        istate = "0"
        ostate = "16"
        samples = ["39","39"]
        threads = ["128","128"]

path = "../circuits/"+circuit+"/ExecutionData/"

algorithms = ["IS","BD_MIS"]

files = [path + circuit + ".data_stats_" + alg + "_V_SP_" + istate + "_" + ostate + "_" + Nsamples + "_" + Nthreads + "_average.csv" for alg,Nsamples,Nthreads in zip(algorithms,samples,threads)]

#print (files)
```

```python
# Read CSVs

data = {}
for ndxfile, file in enumerate(files):
    alg = algorithms[ndxfile] 
    data[alg] = {}
    #print ('Alg is ', alg, '; file is ', file)
    with open(file) as csvfile:
        csvreader = csv.reader(csvfile, delimiter=',')
        for rowndx, row in enumerate(csvreader):
            #print (row)
            if rowndx==0: continue     # 1st row is header
            elif rowndx==1:
                true_w = complex(float(row[0]),float(row[1]))
                data[alg]["true_w"] = true_w  
                #print ("True w = ", true_w)
            elif rowndx==2:     # 3rd row is header
                dataTitles = [s.strip() for s in row]
                dataTitles.remove("RvarianceR")
                dataTitles.remove("RvarianceI")
                # add 3 more attributes
                #dataTitles.append ("variance")
                #dataTitles.append ("L2")
                dataTitles.append ("efficiency")
                for title in dataTitles:
                    data[alg][title] = [] 
                #print(dataTitles)
            else:
                #print (row)
                for titlendx, title in enumerate(dataTitles):
                    if titlendx <10:
                        data[alg][title].append(float(row[titlendx])) 
                    elif titlendx == 10:   # efficiency
                        time = (float(row[1]))
                        variance = (float(row[7]))
                        if variance > 0.:
                            efficiency = 1./(time*variance)
                        else:
                            efficiency = sys.float_info.max
                        data[alg]["efficiency"].append(efficiency)
    #print (dataTitles)
    
```

```python
## Plot allows specification of dependent and in dependent variables
# choose each from
'n_samples', 'time', 'n_paths', 'n_NZero_paths', 'estimateR', 'estimateI', 'varianceR', 'varianceI', 'var', 'L2', 'efficiency'
# independent variable
X_var = 'n_paths'
# dependent variable
Y_var = 'var'

# log X axis ?
log_X_axis = True
# log Y axis ?
log_Y_axis = True
```

```python
fig, ax = plt.subplots()

if log_X_axis:
    # convert x-axis to Logarithmic scale
    plt.xscale("log")
else:
    plt.xscale("linear")


if log_Y_axis:
    # convert y-axis to Logarithmic scale
    plt.yscale("log")
else:
    plt.yscale("linear")

# for each algorithm
for alg in algorithms:
    # Data for plotting
    X = data[alg][X_var]
    Y = data[alg][Y_var]

    ax.plot(X, Y, label = alg)

ax.set(xlabel=X_var, ylabel=Y_var,
       title='Circuit ' +  circuit)
ax.grid()
ax.legend ()


plt.show()
```

```python
img = path + "/circuit_"+circuit + "_" + alg + "_" + istate + "_" + ostate + "_" + X_var + "_VS_" + Y_var + ".png"
fig.savefig(img)
```
